```python
import pandas as pd

# 1. 定义文件路径（注意使用 r'' 防止转义字符问题）
file_path = r'D:\BaiduNetdiskDownload\Quants\tdx\tdx_signal\test_signal.txt'

# 2. 读取文件并解析
# sep='\s+' 表示使用正则匹配一个或多个空白字符（兼容制表符和空格）
# header=None 表示文本中没有列名，需要手动指定
df = pd.read_csv(
    file_path, 
    sep='\s+', 
    header=None, 
    encoding='gbk',
    names=['股票代码', '股票名称', '时间', '价格', '涨跌幅', '数值', '信号']
)

# 3. 打印查看结果
print("成功读取信号数据：")
print(df)
```
