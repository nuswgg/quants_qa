```python
import pymongo

client = pymongo.MongoClient('mongodb://localhost:27017/')
db = client['quantaxis']  # 请替换为您实际的数据库名

# 查询最大日期
result = db.stock_day.find_one(
    {'code': '000002'},
    sort=[('date', pymongo.DESCENDING)]
)

if result:
    print("最大日期:", result['date'])
else:
    print("未找到该股票的记录")
```
