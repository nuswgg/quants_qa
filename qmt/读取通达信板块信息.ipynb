```python
from xtquant import xtdata

# 1. 定义板块代码和周期
# 注意：通达信行业板块在 xtquant 中通常带有后缀，具体后缀取决于你的数据源配置
# 常见的后缀为 .TDX，如果不确定，可以先通过 xtdata.get_stock_list_in_sector("沪深指数") 查看
stock_code = "881002.TDX"  
period = "1d"  # 日线数据

# 2. 下载历史数据到本地缓存
# incrementally=True 表示增量下载，节省时间
print(f"正在下载 {stock_code} 的日线数据...")
xtdata.download_history_data(stock_code, period=period, incrementally=True)

# 3. 从本地读取历史日线数据
# field_list 指定需要的字段，count=-1 表示获取所有已下载的数据
data = xtdata.get_market_data_ex(
    field_list=['open', 'high', 'low', 'close', 'volume'],
    stock_list=[stock_code],
    period=period,
    count=-1
)

# 4. 打印或处理数据
if stock_code in data and not data[stock_code].empty:
    df = data[stock_code]
    print(f"成功获取 {len(df)} 条日线数据：")
    print(df.tail(10))  # 打印最后10条数据查看
else:
    print("未获取到数据，请检查板块代码后缀是否正确（如尝试 881002.SH 或 881002.SZ）")
```
