```python
# coding:gbk

# 1. 定义一个全局对象作为存储容器
class G:
    pass

g = G()

def init(ContextInfo):
    # 2. 在初始化函数中，为 g 对象添加自定义属性
    g.stock_pool = ['000001.SZ', '600000.SH']  # 股票池
    g.position = 0                             # 持仓状态
    g.last_trade_price = 0                     # 上次交易价格
    g.some_counter = 0                         # 计数器
    print("全局变量已初始化")

def handlebar(ContextInfo):
    # 3. 在核心逻辑中，安全地读取和修改全局属性
    current_bar_index = ContextInfo.barpos
    print(f"处理第 {current_bar_index} 根K线, 当前持仓: {g.position}")

    # 模拟逻辑：更新属性
    g.some_counter += 1
    if g.some_counter % 10 == 0:
        g.position = 100
        print(f"计数器到达10的倍数，更新持仓为 {g.position}")
```
