```python
from xtquant.xttrader import XtQuantTrader
from xtquant.xttype import StockAccount
import random

# 1. 指定本地 miniQMT 客户端路径（请替换为你电脑上的实际路径）
# 注意：路径必须指向 userdata_mini 文件夹
miniqmt_path = r'D:\国金QMT交易端模拟\userdata_mini'

# 2. 设置会话 ID（个人使用可随意设置，若多开需保证唯一）
session_id = int(random.randint(100000, 999999))

# 3. 实例化交易连接对象
trader = XtQuantTrader(miniqmt_path, session_id)

# 4. 启动交易线程并建立连接
trader.start()
connect_result = trader.connect()

# 5. 检查连接状态（返回 0 表示连接成功）
if connect_result == 0:
    print('miniQMT 连接成功！')
else:
    print('连接失败，请检查 miniQMT 是否已启动极简模式')
    trader.stop()
    exit()

# 6. 订阅资金账户（必须订阅才能接收推送和进行后续操作）
account = StockAccount('10503139')
subscribe_result = trader.subscribe(account)

if subscribe_result == 0:
    print('账户订阅成功！')
else:
    print('账户订阅失败！')
```
