```python
from pytdx.exhq import TdxExHq_API
from pytdx.util.best_ip import select_best_ip

# 1. 自动测试并选择最优的扩展行情服务器
print("正在自动寻找最优的扩展行情服务器...")
try:
    best_ip_info = select_best_ip()
    best_ip = best_ip_info['ip']
    best_port = best_ip_info['port']
    print(f"✅ 找到最优服务器: {best_ip}:{best_port}")
except Exception as e:
    print(f"❌ 自动选择服务器失败: {e}")
    # 如果自动选择失败，可以手动指定一个扩展行情服务器作为备用
    best_ip, best_port = '112.74.214.43', 7727 

# 2. 连接服务器并获取数据
api = TdxExHq_API()
if api.connect(best_ip, best_port):
    print("✅ 成功连接到扩展行情服务器！")
    try:
        # 获取市场代码为 70（扩展板块指数）的品种列表
        instruments = api.get_instrument_info(70, 0, 100)
        
        # 增加判断，防止返回 None 导致报错
        if instruments:
            semi_indices = [item for item in instruments if '半导体' in item.get('name', '')]
            print("\n申万半导体相关指数信息：")
            for index_info in semi_indices:
                print(f"代码: {index_info['code']}, 名称: {index_info['name']}")
        else:
            print("⚠️ 未获取到扩展板块指数数据，当前服务器可能不支持该接口。")
            
    except Exception as e:
        print(f"❌ 获取数据时发生错误: {e}")
    finally:
        api.disconnect()
        print("已断开服务器连接。")
else:
    print("❌ 连接扩展行情服务器失败，请检查网络环境。")
```
